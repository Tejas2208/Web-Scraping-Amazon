# Univariate Analysis

## Univariate Analysis – Brand Distribution
Analyzing frequency of sneaker brands to identify which brand dominates the dataset.

```python
top_brands = df['Title'].value_counts().head(15)

plt.figure(figsize=(12,7))

sns.barplot(
    x=top_brands.values,
    y=top_brands.index,
    palette="viridis"
)

plt.title("Top 15 Sneaker Brands on Amazon",weight='bold')
plt.xlabel("Number of Products")
plt.ylabel("Brand")

plt.tight_layout()
plt.show()
```

## Observations:
- PUMA has the highest count in the dataset more than 60.
- The distribution is uneven, with some brands contributing significantly more products than others.

## Business Insight:  
- Customers have more choices among leading brands.
- Smaller brands may need better visibility or marketing to compete.

2. Most discounts lie between 35% to nearly 60%. Anything outside this range is either under-discounted or aggresively pushed.
3. There are no strong outliers visible. Discounts are consistent, not dominated by extreme outliers.This tells that discounting standard pricing strategy across Sneakers rather than any promotional tactic.

## Business Insight:  
- Large discounts may indicate promotional campaigns or inventory clearance.
- Discount strategies differ significantly across brands.

# Bivariate Analysis

## Bivariate Analysis – Price vs Rating
Examining relationship between product price and customer ratings to identify if higher-priced products receive better ratings.

```python
# Price vs Rating
plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x="Original Price",
    y="Rating",
    alpha=0.7,
    s=80
)

sns.regplot(
    data=df,
    x="Original Price",
    y="Rating",
    scatter=False,
    color="red"
)

plt.title("Relationship between Price and Rating",weight='bold')
plt.savefig("price and rating.png", dpi=300, bbox_inches="tight")
plt.show()
```

## Observations:
- Points are spread randomly across the graph, that means there's no strong relationship between 'Price' and 'Rating'.
- Maximum Ratings are stucked bewtween 3.8 to 4.3 even the Sneaker is cheap or expensive, this tells that price has negligible impact on Ratings.
- Expensive Sneaker Brands are not consistently better, you can clearly see that Expensive ones has average Ratings where Mid price Sneakers have High Ratings.
- The scatter plot shows no strong relationship bewtween Price and Ratings, indicating that higher priced sneakers do not necessarily receive better customer ratings.

## Business Insight:
- Increasing price alone does not improve customer satisfaction.
- Product quality and customer experience are more important than price

```python
# Highest Rated Brands (Minimum 5 Products)

brand_rating = (
    df.groupby("Title")
      .agg(
          Average_Rating=("Rating", "mean"),
          Product_Count=("Title", "count")
      )
)

# Keep brands with at least 5 products
brand_rating = brand_rating[brand_rating["Product_Count"] >= 5]

# Sort by average rating
brand_rating = (
    brand_rating
    .sort_values("Average_Rating", ascending=False)
    .head(10)
)

plt.figure(figsize=(12,6))

ax = sns.barplot(
    data=brand_rating,
    x="Average_Rating",
    y=brand_rating.index,
    palette="crest"
)

# Add rating values at the end of each bar
for i, value in enumerate(brand_rating["Average_Rating"]):
    ax.text(
        value + 0.01,
        i,
        f"{value:.2f}",
        va="center",
        fontsize=11,
        fontweight="bold"
    )

plt.title(
    "Top 10 Highest Rated Sneaker Brands",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel("Average Rating")
plt.ylabel("Brand")

plt.xlim(4, 5)

sns.despine()

plt.tight_layout()

plt.show()
```

## Observations:
- Several brands maintain consistently high customer ratings.
- Filtering brands by product count produces more reliable results.
## Business Insight:
- Consistently high-rated brands build stronger customer trust.
- These brands may justify premium pricing.

# Multivariate Analysis

## Correlation Heatmap Analysis

This heatmap shows the correlation between numerical features such as:
- Discounted Price
- Original Price
- Rating
- Review Count
- Discount %

```python
plt.figure(figsize=(8,6))

corr = df[
    ['Discounted Price',
     'Original Price',
     'Rating',
     'Review Count',
     'Discount %']
].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    square=True
)

plt.title("Correlation Matrix",weight='bold')
plt.savefig("correlation_matrix.png", dpi=300, bbox_inches="tight")

plt.show()
```

## Observation:  
- Original Price and Discounted Price show a strong positive correlation.
- Rating has little correlation with price.
- Review Count is only weakly correlated with other variables.
## Business Insight:
- Pricing is largely driven by the original price.
- Customer ratings appear to depend on product quality rather than pricing.
